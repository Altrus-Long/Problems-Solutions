```python
# HJ75 最长公共子串

# 方法1：广义后缀树
class GeneralizedSuffixTrie:
    class Node:
        __slots__ = ('children', 'string_ids')
        def __init__(self):
            self.children = {}          # 字符 -> 子节点
            self.string_ids = set()     # 该节点子树中包含的字符串编号

    def __init__(self, strings):
        self.strings = strings
        self.root = self.Node()
        self._build()

    def _build(self):
        """将所有字符串的所有后缀插入 Trie"""
        for sid, s in enumerate(self.strings):
            for i in range(len(s)):
                node = self.root
                for j in range(i, len(s)):
                    ch = s[j]
                    if ch not in node.children:
                        node.children[ch] = self.Node()
                    node = node.children[ch]
                # 后缀末尾节点，记录来源字符串编号
                node.string_ids.add(sid)

        # 后序遍历，合并子节点的字符串集合
        self._collect_string_ids(self.root)

    def _collect_string_ids(self, node):
        for child in node.children.values():
            self._collect_string_ids(child)
            node.string_ids |= child.string_ids

    def longest_common_substring(self):
        """返回所有字符串的最长公共子串"""
        num_strings = len(self.strings)
        best = ""
        best_len = 0

        def dfs(node, depth, path):
            nonlocal best, best_len
            # 如果当前节点包含所有字符串，尝试更新答案
            if len(node.string_ids) == num_strings:
                if depth > best_len:
                    best_len = depth
                    best = ''.join(path)

            for ch, child in node.children.items():
                path.append(ch)
                dfs(child, depth + 1, path)
                path.pop()

        dfs(self.root, 0, [])
        return best


# 测试
if __name__ == "__main__":
    strings = ["abcpqrabc", "xabcpqr", "zzabcpqr"]
    gst = GeneralizedSuffixTrie(strings)
    result = gst.longest_common_substring()
    print("最长公共子串:", result)  # 输出: abcpqr
```

```python
# 方法2，动态规划
s = 'awaabb'
t = 'aawbb'
d = [[0]*(len(t)+1) for _ in range(len(s)+1)]
ans = 0
for i in range(1,len(s)+1):
    for j in range(1,len(t)+1):
        if s[i-1] == t[j-1]:
            d[i][j] = d[i-1][j-1]+1
        else:
            d[i][j] = 0
        ans = max(ans,d[i][j])
print(ans)
```

```python
# HJ85 最长回文子串3种解法。
# https://leetcode.cn/problems/longest-palindromic-substring/solutions/255195/zui-chang-hui-wen-zi-chuan-by-leetcode-solution/
def longestPalindrome1(s: str) -> str:
    # 动态规划
    # 状态转移方程 d[i][j]表示s[i...j]是否为回文子串，用bool值即可
    # d[i][j] = d[i+1][j-1]^s[i]==s[j]
    # d[i][i]表示s[i],长度为1必为回文子串，设为True
    # d[i][i+1] = s[i]==s[i+1]
    n = len(s)
    if n < 2:
        return s
    max_len = 1
    begin = 0
    dp = [[False] * n for _ in range(n)] # dp[i][j] 表示 s[i..j] 是否是回文串
    
    for i in range(n):  # 长度为1
        dp[i][i] = True
    
    # 递推开始
    # 先枚举子串长度
    for L in range(2, n + 1):
        # 枚举左边界，左边界的上限设置可以宽松一些
        for i in range(n):
            # 由 L 和 i 可以确定右边界，即 j - i + 1 = L 得
            j = L + i - 1
            # 如果右边界越界，就可以退出当前循环
            if j >= n:
                break
                
            if s[i] != s[j]:
                dp[i][j] = False 
            else:
                if j - i < 3:
                    dp[i][j] = True
                else:
                    dp[i][j] = dp[i + 1][j - 1]
            
            # 只要 dp[i][j] == true 成立，就表示子串 s[i..j] 是回文，此时记录回文长度和起始位置
            if dp[i][j] and L > max_len:
                max_len = L
                begin = i
    return s[begin:begin + max_len]

def longestPalindrome2(s:str)->str:
    def expandAroundCenter(s, left, right):
        while left >= 0 and right < len(s) and s[left] == s[right]:
            left -= 1
            right += 1
        return left + 1, right - 1
    start, end = 0, 0
    for i in range(len(s)):
        left1, right1 = expandAroundCenter(s, i, i)
        left2, right2 = expandAroundCenter(s, i, i + 1)
        if right1 - left1 > end - start:
            start, end = left1, right1
        if right2 - left2 > end - start:
            start, end = left2, right2
    return s[start: end + 1]



def longestPalindrome3(s: str) -> str:
    # Manacher 算法
    if not s:
        return ""
    # 1. 插入分隔符，统一奇偶长度
    # 例如 "abba" -> "^#a#b#b#a#$"
    t = "^#" + "#".join(s) + "#$"
    n = len(t)
    p = [0] * n          # p[i] 表示以 t[i] 为中心的回文半径
    center = 0           # 当前最右回文的中心
    right = 0            # 当前最右回文的右边界
    max_len = 0          # 最长回文半径
    max_center = 0       # 最长回文的中心

    for i in range(1, n - 1):
        # 2. 利用对称性初始化 p[i]
        if i < right:
            mirror = 2 * center - i
            p[i] = min(right - i, p[mirror])

        # 3. 中心扩展
        while t[i + p[i] + 1] == t[i - p[i] - 1]:
            p[i] += 1

        # 4. 更新最右回文边界
        if i + p[i] > right:
            center = i
            right = i + p[i]

        # 5. 记录最长回文
        if p[i] > max_len:
            max_len = p[i]
            max_center = i

    # 6. 从处理后的字符串还原原串下标
    start = (max_center - max_len) // 2
    return s[start:start + max_len]


# 测试
if __name__ == "__main__":
    s = 'qhbrivaighqmgafhthxicdiixpefhwwefdebwczswqqdjxulhuhceqrxechddtlbbltddhcexrqechuhluxjdqqwszcwenakceymkxfqpqxctbsousrwwhooxjtcqnvb'
    s='abba'
    print(len(longestPalindrome1(s)))
    print(len(longestPalindrome2(s)))
    print(len(longestPalindrome3(s)))
```

```python
# Hj87 密码强度等级

password = '38$@NoNoN'
score = 0

length = 0
uppercount = 0
lowercount = 0
digitcount = 0
othercount = 0
for c in password:
    length +=1
    if 'A'<=c<='Z':
        uppercount +=1
    if 'a'<=c<='z':
        lowercount+=1
    if '0'<=c<='9':
        digitcount+=1
    if 0x21<=ord(c)<=0x2f \
    or 0x3A<=ord(c)<=0x40\
    or 0x5B<=ord(c)<=0x60\
    or 0x7B<=ord(c)<=0x7E:
        othercount+=1
if length<=4:
    score+=5
elif 5<=length<=7:
    score+=10
else:
    score+=25

if uppercount + lowercount == 0:
    pass
elif (uppercount>0 and lowercount==0) or (uppercount==0 and lowercount>0):
    score+=10
else:
    score+=20

if digitcount == 0:
    pass
elif digitcount==1:
    score+=10
else:
    score+=20

if othercount ==0:
    pass
elif othercount==1:
    score+=10
else:
    score+=25

if uppercount>0 and lowercount>0 and digitcount>0 and othercount>0:
    score+=5
elif (uppercount+lowercount)>0 and digitcount>0 and othercount>0:
    score+=3
elif (uppercount+lowercount)>0 and digitcount>0:
    score+=2

if score>=90:
    print('VERY_SECURE')
elif score>=80:
    print('SECURE')
elif score>=70:
    print("VERY_STRONG")
elif score>=60:
    print("STRONG")
elif score>=50:
    print('AVERAGE')
elif score>=25:
    print("WEAK")
else:
    print("VERY_WEAK")
```
