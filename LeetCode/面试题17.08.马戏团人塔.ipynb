```python
import bisect
from typing import List
class Solution:
    def bestSeqAtIndex(self, height: List[int], weight: List[int]) -> int:       
        dp=[]
        for a,b in sorted(zip(height,weight),key = lambda x:(x[0],-x[1])):
            # 身高升序，身高相同，体重降序
            pos = bisect.bisect_left(dp,b)
            dp[pos:pos+1] = [b]
        print(dp)
        return len(dp)

Solution().bestSeqAtIndex(height = [65,70,56,75,60,60,60,63,68] ,
                        weight = [100,150,90,190,95,91,90,120,110])
```
