```python
from typing import List


class Solution:
    def maxDepthAfterSplit(self, seq: str) -> List[int]:
        ans = []
        d = 0
        for c in seq:
            if c == '(':
                d += 1
                ans.append(d % 2)
            if c == ')':
                ans.append(d % 2)
                d -= 1
        return ans
```

```python
class Solution:
    def maxDepthAfterSplit(self, seq: str) -> List[int]:
        ans=[]
        a=b=0 #这是深度，初始都是0
        for s in seq:
            if s=='(':
                if a<=b:
                    a+=1
                    ans.append(0)
                else:
                    b+=1
                    ans.append(1)
            elif s==')':
                if a>b:
                    a-=1
                    ans.append(0)
                else:
                    b-=1
                    ans.append(1)
        return ans
```
