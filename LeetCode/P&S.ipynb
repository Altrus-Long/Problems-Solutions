```python
# 素数判断
# 试除法
import math
def is_prime(n: int) -> bool:
    if n <= 1:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False
    # 从 3 开始，只检查奇数因子，直到 √n
    for i in range(3, int(math.isqrt(n)) + 1, 2):
        if n % i == 0:
            return False
    return True
%time is_prime(9)
```
