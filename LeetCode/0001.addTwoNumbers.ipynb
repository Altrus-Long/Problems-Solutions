```python
# 两数相加

# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
class Solution:
    def addTwoNumbers(self, l1: ListNode | None, l2: ListNode | None) -> ListNode | None:
        # ListNode只有init方法，因此每次构造节点时需要先得到当前要构造节点的val值。
        # 先构造头节点，因为最后需要返回的是头节点。
        # 计算个位时，默认的进位为0
        carry = 0
        d = l1.val+l2.val+carry
        head = ListNode(d%10)
        carry = d//10

        current = head #设定游标节点
        l1 = l1.next#进入十位
        l2 = l2.next#进入十位

        while l1 and l2:# 均为非空
            d = l1.val+l2.val+carry
            current.next = ListNode(d%10)
            carry = d//10
            current = current.next
            l1 = l1.next 
            l2 = l2.next
        # l1或者l2一个为空，跳出循环，空的链表不再参与计算
        while l1:
            # if carry == 0:#无进位，剩余部分接到current之后即可，若l1、l2长度相差悬殊，可以启用节省时间
            #     current.next = l1
            #     break
            d = l1.val+carry
            current.next = ListNode(d%10)
            carry = d//10
            current = current.next
            l1 = l1.next
        while l2:
            # if carry == 0:#无进位，剩余部分接到current之后即可
            #     current.next = l2
            #     break
            d = l2.val+carry
            current.next = ListNode(d%10)
            carry = d//10
            current = current.next
            l2 = l2.next 
        if carry > 0: # 最高位还需要进1，而此时l1，l2都空
            current.next = ListNode(carry)
        return head
```

```python
l1 = ListNode(2,ListNode(4,ListNode(3))) #342
l2 = ListNode(5,ListNode(6,ListNode(4))) #465
res = Solution().addTwoNumbers(l1,l2) #807
while res:
    print(res.val,end='')
    res = res.next
```
