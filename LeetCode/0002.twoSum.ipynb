```python
# 两数之和（LeetCode）
class Solution:
    def twoSum(self, nums: list[int], target: int) -> list[int]:
        hashtable = dict()
        for i in range(len(nums)):
            need = target-nums[i]
            if need in hashtable:
                return [hashtable[need],i]
            hashtable[nums[i]] = i
        return []
```

```python
Solution().twoSum([2,7,11,15],9)
```
