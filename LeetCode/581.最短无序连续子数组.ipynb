```python
class Solution:
    def findUnsortedSubarray(self, nums: list[int]) -> int:
        n = len(nums)
        maxn,right = float('-inf'),-1
        minx,left = float('inf'),-1

        for i in range(n):
            if maxn > nums[i]:
                right = i
            else:
                maxn = nums[i]

            if minn < nums[n-i-1]:
                left = n-i-1
            else:
                minn= nums[n-i-1]
        return 0 if right == -1 else right - left +1
                    
    Solution().findUnsortedSubarray([2,6,4,8,10,9,15])
```
