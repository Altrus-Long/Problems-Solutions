```python
class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        occ = set()
        n = len(s)
        rk,ans = 0,0
        for i in range(n):
            while rk < n and s[rk] not in occ:
                occ.add(s[rk])
                rk+=1
            ans = max(ans,rk-i)
            occ.remove(s[i])
        return ans

s =  "abcabcbb"
Solution().lengthOfLongestSubstring(s)
```
