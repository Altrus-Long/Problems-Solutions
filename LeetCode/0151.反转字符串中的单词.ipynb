```python
class Solution:
    def reverseWords(self, s: str) -> str:
        return ' '.join(reversed(s.split())).strip()
s = 'the   sky is blue'
Solution().reverseWords(s)
```


# strip（）
strip() —— 去除首尾字符，默认去掉首尾的：空格、\t、\n、\r 等空白字符。

指定要去掉的字符：print(s.strip("*"))   # hello

strip(chars) 的参数是字符集合，不是完整子串：
s = "abcxyzabc"
print(s.strip("abc"))  # xyz

只去一边
s = "  hello  "
print(s.lstrip())  # "hello  "
print(s.rstrip())  # "  hello"

# split() —— 拆分成列表
不传参数时，按任意空白字符分割。
连续多个空格视为一个分隔符。
自动忽略首尾空白。

指定分隔符
s = "a,b,c"
print(s.split(","))


限制分割次数
s = "a,b,c,d"
print(s.split(",", 2))

从右边分割
s = "a,b,c,d"
print(s.rsplit(",", 2))


按行分割
s = "line1\nline2\r\nline3"
print(s.splitlines())
